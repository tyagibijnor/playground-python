#### We have multiple option to clean data.
#### We can drop missing value data with dropna()
#### We can use Mean / Median Imputation to fill missing data with mean / meadian value. 

```python
df['Heart Rate (BPM)'].describe()
```

```python
df['Heart Rate (BPM)'].median()
```

```python
df['Heart Rate (BPM)'].mean()
```

```python
df['Heart Rate (BPM)'].fillna(df['Heart Rate (BPM)'].mean())
```

## moving to make it permanent for 'Heart Rate BPM'

```python
df['Heart Rate (BPM)']=df['Heart Rate (BPM)'].fillna(df['Heart Rate (BPM)'].mean())
```

```python
df['Heart Rate (BPM)'].isnull().sum()
```

```python
### now we have 10 thousand rows now
## drpping USERID columns with using columns drop

df=df.drop(columns=['User ID'])
```

```python
## checking for True null value
df[df['Blood Oxygen Level (%)'].isnull()==True]
```

```python
## checked values for Blood Oxygen Level (%) and filling with mean value
df['Blood Oxygen Level (%)'].mean()
```

```python
## fill value to user Forward fill or Backward fill
## like time series type of data
## in the case of Forward filling for example if 1 1 2 3 5 NaN 5 8 6 
## then 5 will fill on place NaN
## And for Backward NaN will changed with 5
## method --- ffill and bfill

## df['Step Count']=df['Step Count'].fillna(method='ffill')

## Note for Method : Your code uses fillna(method='ffill'). 
# In the environment tested, pandas is 3.1.0, 
# where fillna() no longer accepts method.
df["Step Count"] = df["Step Count"].ffill()
```
