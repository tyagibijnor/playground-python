```python
## looks all have values now
## test for Simple Imputation
## reloading csv
## and will fill using import SimpleImputer

from sklearn.impute import  SimpleImputer

csv = pd.read_csv('unclean_smartwatch_health_data.csv')

csv.info()
```

```python
## using strategy to median
## and fill value with fit_transform to mean/median
## and using ravel() to change 2D
imputer=SimpleImputer(strategy='median')
csv['Heart Rate (BPM)']=imputer.fit_transform(csv[['Heart Rate (BPM)']]).ravel()
```

```python
## after fixing Heart Rate (BPM) checking info
csv.info()
```

```python
## that is fixed now
csv['Blood Oxygen Level (%)'].isnull().sum()
imputer=SimpleImputer(strategy='mean')
csv['Blood Oxygen Level (%)']=imputer.fit_transform(csv[['Blood Oxygen Level (%)']]).ravel()
csv['Step Count']=imputer.fit_transform(csv[['Step Count']]).ravel()
```

```python
## after fixing Blood Oxygen Level (%) checking info
csv.info()
```

```python
## this Blood Oxygen Level (%) is fixed now
## Activity and Stree Level is categorical column
csv['Stress Level'].isnull().sum()
```

```python
## using most frequest value to fill missing data
imputer=SimpleImputer(strategy='most_frequent')
csv['Stress Level']=imputer.fit_transform(csv[['Stress Level']]).ravel()
```
